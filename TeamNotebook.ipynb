## Section 2: Core DES Helper Functions
Low-level building blocks used by the DES algorithm.

```python
def permute(block, table):
    """
    Apply a permutation table to a bit-string.
    Table indices are 1-based (standard DES convention).
    """
    return ''.join(block[i - 1] for i in table)


def xor_bits(a, b):
    """Bitwise XOR of two equal-length bit-strings."""
    return ''.join('0' if x == y else '1' for x, y in zip(a, b))


def s_box_substitution(bits_48):
    """
    Apply all 8 S-Boxes to a 48-bit string, producing a 32-bit output.
    For each 6-bit chunk:
      row = first bit concatenated with last bit  (2-bit value)
      col = middle four bits                       (4-bit value)
    """
    result = ''
    for i in range(8):
        chunk = bits_48[i * 6:(i + 1) * 6]   # 6 bits for S-box i
        row = int(chunk[0] + chunk[5], 2)     # bits 0 and 5
        col = int(chunk[1:5], 2)              # bits 1-4
        val = S_BOXES[i][row][col]
        result += format(val, '04b')          # 4-bit binary output
    return result


def feistel_function(R, K):
    """
    DES Feistel (f) function:
      1. Expand R from 32 to 48 bits using table E
      2. XOR with round key K (48 bits)
      3. Apply 8 S-Boxes → 32 bits
      4. Apply permutation P → 32 bits
    Returns all intermediate values for display purposes.
    """
    expanded   = permute(R, E)                 # 32 → 48 bits
    xored      = xor_bits(expanded, K)         # key mixing
    substituted = s_box_substitution(xored)    # 48 → 32 bits
    permuted   = permute(substituted, P)       # final permutation
    return expanded, xored, substituted, permuted


print("Helper functions defined: permute, xor_bits, s_box_substitution, feistel_function")
```

## Section 3: DES Encrypt / Decrypt (Single 64-bit Block)
The core DES block cipher, with step-by-step output per round.

```python
def des_encrypt_block(block_64, round_keys, label="DES Encryption"):
    """
    Encrypt a 64-bit block using DES with the given 16 round keys.
    Always prints every step clearly.
    Steps:
      1. Initial Permutation (IP)
      2. 16 Feistel rounds
      3. Swap halves
      4. Final Permutation (FP)
    """
    # ── Step 1: Initial Permutation ──────────────────────────────
    after_ip = permute(block_64, IP)
    L = after_ip[:32]
    R = after_ip[32:]

    print(f"\n{'═'*65}")
    print(f"  {label}")
    print(f"{'═'*65}")
    print(f"  Input (hex) : 0x{int(block_64, 2):016x}")
    print(f"  After IP    : {after_ip}")
    print(f"  L0          : {L}")
    print(f"  R0          : {R}")
    print(f"  {'─'*61}")
    print(f"  {'Round':<8} {'Expansion (E)':<50}")
    print(f"  {'':8} {'Key XOR':<50}")
    print(f"  {'':8} {'S-Box Out':<50}")
    print(f"  {'':8} {'P-Box Out':<50}")
    print(f"  {'':8} {'New L':<50}")
    print(f"  {'':8} {'New R':<50}")
    print(f"  {'─'*61}")

    # ── Step 2: 16 Feistel Rounds ────────────────────────────────
    for i in range(16):
        expanded, xored, substituted, permuted = feistel_function(R, round_keys[i])
        new_R = xor_bits(L, permuted)   # new right = old left XOR f(R, K)
        new_L = R                       # new left  = old right

        print(f"\n  Round {i+1:02d}")
        print(f"    Expansion (E)    : {expanded}")
        print(f"    Key mixing (XOR) : {xored}")
        print(f"    S-Box output     : {substituted}")
        print(f"    P-Box output     : {permuted}")
        print(f"    L{i+1:<2}              : {new_L}")
        print(f"    R{i+1:<2}              : {new_R}")

        L, R = new_L, new_R

    # ── Step 3: Swap halves ──────────────────────────────────────
    combined = R + L

    # ── Step 4: Final Permutation ────────────────────────────────
    ciphertext = permute(combined, FP)

    print(f"\n  {'─'*61}")
    print(f"  After swap  : {combined}")
    print(f"  After FP    : {ciphertext}")
    print(f"  Output (hex): 0x{int(ciphertext, 2):016x}")
    print(f"  {'═'*61}")

    return ciphertext


def des_decrypt_block(block_64, round_keys, label="DES Decryption"):
    """
    Decrypt a 64-bit block using DES.
    Decryption reuses the same DES structure but applies round keys in REVERSE order.
    """
    return des_encrypt_block(block_64, round_keys[::-1], label=label)


print("DES encrypt/decrypt block functions defined.")
```

## Section 4: Triple DES (EDE) on a Single 64-bit Block
3DES uses three passes over a single block:
1. **Encrypt** with K1
2. **Decrypt** with K2
3. **Encrypt** with K3

```python
def triple_des_block(block_64, k1_keys, k2_keys, k3_keys):
    """
    Triple DES (EDE) on a single 64-bit block.
    EDE = Encrypt(K1) → Decrypt(K2) → Encrypt(K3)
    Always prints every step of all 3 stages.
    """
    print("\n" + "█"*65)
    print("  TRIPLE DES (EDE) — Single 64-bit Block")
    print("█"*65)
    print(f"  Input  (hex): 0x{int(block_64, 2):016x}")

    # Stage 1 — Encrypt with K1
    stage1 = des_encrypt_block(block_64, k1_keys, label="Stage 1 — Encrypt with K1")

    # Stage 2 — Decrypt with K2
    stage2 = des_decrypt_block(stage1,   k2_keys, label="Stage 2 — Decrypt with K2")

    # Stage 3 — Encrypt with K3
    stage3 = des_encrypt_block(stage2,   k3_keys, label="Stage 3 — Encrypt with K3")

    print("\n" + "█"*65)
    print("  3DES SUMMARY")
    print("█"*65)
    print(f"  Input          : 0x{int(block_64, 2):016x}")
    print(f"  After E(K1)    : 0x{int(stage1, 2):016x}")
    print(f"  After D(K2)    : 0x{int(stage2, 2):016x}")
    print(f"  After E(K3)    : 0x{int(stage3, 2):016x}  ← 3DES output")
    print("█"*65)

    return stage3


print("triple_des_block() function defined.")
```

## Section 5: Encrypting a Full Plaintext Message
The plaintext `"Hello 3DES Encryption"` is split into 64-bit (8-byte) blocks with zero-padding on the last block.

```python
def text_to_64bit_blocks(text):
    """
    Convert a UTF-8 string into a list of 64-bit (8-byte) blocks.
    The last block is zero-padded if the text length is not a multiple of 8.
    Returns:
      blocks        — list of 64-character binary strings
      padding_bytes — number of zero bytes added
    """
    byte_data    = text.encode('utf-8')
    padded_len   = ((len(byte_data) + 7) // 8) * 8   # round up to multiple of 8
    padded       = byte_data.ljust(padded_len, b'\x00')
    padding_bytes = padded_len - len(byte_data)

    blocks = []
    for i in range(0, len(padded), 8):
        chunk = padded[i:i + 8]
        bits  = ''.join(format(b, '08b') for b in chunk)
        blocks.append(bits)

    return blocks, padding_bytes


# ── Prepare blocks ────────────────────────────────────────────
blocks, pad_bytes = text_to_64bit_blocks(plaintext)

print(f"Plaintext     : '{plaintext}'")
print(f"Length        : {len(plaintext)} characters / {len(plaintext.encode())} bytes")
print(f"Padding added : {pad_bytes} zero byte(s)")
print(f"Total blocks  : {len(blocks)}")
print()
for i, b in enumerate(blocks):
    text_repr = int(b, 2).to_bytes(8, 'big').rstrip(b'\x00').decode('utf-8', errors='replace')
    print(f"  Block {i+1}: 0x{int(b, 2):016x}  → '{text_repr}'")
```

```python
# ── Encrypt all blocks with 3DES ──────────────────────────────
print("="*65)
print("  3DES-EDE Encryption of All Blocks")
print("="*65)

ciphertext_blocks = []

for idx, block in enumerate(blocks):
    text_repr = int(block, 2).to_bytes(8, 'big').rstrip(b'\x00').decode('utf-8', errors='replace')
    print(f"\n{'█'*65}")
    print(f"  BLOCK {idx+1} — Plaintext: '{text_repr}'  (0x{int(block, 2):016x})")
    print(f"{'█'*65}")

    ct = triple_des_block(
        block,
        first_16_round_keys,
        second_16_round_keys,
        third_16_round_keys
    )
    ciphertext_blocks.append(ct)

print(f"\n{'='*65}")
print("  FINAL SUMMARY — All Ciphertext Blocks")
print(f"{'='*65}")
for i, (pt, ct) in enumerate(zip(blocks, ciphertext_blocks)):
    pt_text = int(pt, 2).to_bytes(8, 'big').rstrip(b'\x00').decode('utf-8', errors='replace')
    print(f"  Block {i+1}: '{pt_text:<10}'  →  0x{int(ct, 2):016x}")
print(f"{'='*65}")
```

## Section 6: Decryption and Verification
To decrypt with 3DES-EDE, the order and direction of operations are reversed:

**Decrypt(K3) → Encrypt(K2) → Decrypt(K1)**

> Note: Only the **input** and **final output** are readable as text.
> The intermediate stages (after D(K3) and after E(K2)) are still scrambled binary data — they will show as `???` because they are not valid text yet.

```python
def bits_to_text(bits):
    """Convert a 64-bit string to a readable text representation.
    Returns '???' for bytes that are not valid printable text."""
    try:
        raw = int(bits, 2).to_bytes(8, 'big')
        text = raw.rstrip(b'\x00').decode('utf-8')
        # check all chars are printable
        if all(c.isprintable() for c in text):
            return f"'{text}'"
        return '??? (encrypted binary — not readable text)'
    except:
        return '??? (encrypted binary — not readable text)'


def triple_des_decrypt_block(block_64, k1_keys, k2_keys, k3_keys):
    """
    Decrypt a 3DES-EDE ciphertext block.
    Inverse of EDE = Decrypt(K3) → Encrypt(K2) → Decrypt(K1)
    """
    d1 = des_decrypt_block(block_64, k3_keys, label="Inv Stage 1 — Decrypt with K3")
    d2 = des_encrypt_block(d1,       k2_keys, label="Inv Stage 2 — Encrypt with K2")
    d3 = des_decrypt_block(d2,       k1_keys, label="Inv Stage 3 — Decrypt with K1")
    return d1, d2, d3


# ── Decrypt all ciphertext blocks with full stage trace ───────
print("="*65)
print("  3DES-EDE Decryption — Stage by Stage")
print("="*65)

recovered_text = ''

for i, ct in enumerate(ciphertext_blocks):
    d1, d2, d3 = triple_des_decrypt_block(
        ct,
        first_16_round_keys,
        second_16_round_keys,
        third_16_round_keys
    )

    pt_bytes = int(d3, 2).to_bytes(8, 'big')
    pt_text  = pt_bytes.rstrip(b'\x00').decode('utf-8')
    recovered_text += pt_text

    print(f"\n  Block {i+1}:")
    print(f"    Ciphertext input  : 0x{int(ct,2):016x}  →  {bits_to_text(ct)}")
    print(f"    After D(K3)       : 0x{int(d1,2):016x}  →  {bits_to_text(d1)}")
    print(f"    After E(K2)       : 0x{int(d2,2):016x}  →  {bits_to_text(d2)}")
    print(f"    After D(K1) ✓     : 0x{int(d3,2):016x}  →  '{pt_text}'  ← recovered text")

print(f"\n{'='*65}")
print(f"  Reassembled plaintext : '{recovered_text}'")
print(f"  Original  plaintext   : '{plaintext}'")
assert recovered_text == plaintext, "ERROR: Decryption did not match original!"
print("\n  ✓ Decryption verified — recovered text matches original plaintext.")
```
